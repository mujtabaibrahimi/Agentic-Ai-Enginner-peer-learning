```python
# nested dictionary
students = {
    "student01": {'name': "Ali Ahmad", 'age': 21, 'city': 'Lahore'},
    "student02": {'name': "Sara Khan", 'age': 22, 'city': 'Karachi'},
    "student03": {'name': "Zainab Ali", 'age': 20, 'city': 'Islamabad'}}

print(students)
```

```python
# dictionary comprehension

square = {x:x*1 for x in range(10)}

print(square)

addition = {x: x+2 for x in square }
print(addition)
```

```python
#  conditional comprehension dictionary

x = {x: x+2 for x in square if x%2==0}

print(x)
```

```python
numbers = {1,2,4,5,5,5,5,6,6,6,6,7,7,7,8,8,8,9,9}
frequency ={}

for number in numbers:
    if number in frequency:
        frequency[number] +=1
    else:
        frequency[number] =1   

print(frequency)
```
