```python
##---Simple Calculator---

num1 = float(input("enter the first number: "))
num2 = float(input("enter the second number: "))

operation= input("enter the operation(+,-,*,/):")

if operation == "+":
    print(num1 + num2)

elif operation == "-":
   print(num1 - num2)

elif operation == "*":
    print(num1 * num2)

elif operation == "/":
    print(num1 / num2)

elif operation !=0:
  print("invalid operation")

else:
 result = 'invalid operation'

 print("Result:", result)
```

```python
# Display a prime number

for num in range(1,101):

    if num > 1:
        for i in range(2, num):
            if (num % i) == 0:
                break

            else:
                print(num)
```

```python
# Access list

fruits = ["apple", "banana", "cherry", "date", "quennow", "Pomengranate"]

# indexing

print((fruits[1:3]))
print(fruits[:-1])

# modifying the list

fruits[1:] = "blueberry"
print(fruits)

for fruit in fruits:
    print(fruit)
```

```python
# organzing students grade

grades = [85, 92, 78, 90, 88]


average_grades = sum(grades) // len(grades)
print(f"Average Grade: {average_grades:.2f}")

highest_grade = max(grades)
lowest_grade = min(grades)
print(f"Highest Grade: {highest_grade}")
print(f"Lowest Grade: {lowest_grade}")


```
